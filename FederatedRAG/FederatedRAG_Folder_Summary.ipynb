### RagBenchEvaluation_Summary.md
Source (taken from file): RagBenchEvaluation_Summary.md
- Focuses on RagBench TechQA with percentile sensitivity analysis for semantic chunking.

- Recommends high-percentile semantic chunking when storage/cost is priority, fixed-size when accuracy is priority.

```python
import json
from pathlib import Path

def notebook_overview(path):
    data = json.loads(Path(path).read_text(encoding='utf-8', errors='ignore'))
    cells = data.get('cells', [])
    code_count = sum(1 for c in cells if c.get('cell_type') == 'code')
    md_count = sum(1 for c in cells if c.get('cell_type') == 'markdown')
    return {
        'file': path,
        'total_cells': len(cells),
        'code_cells': code_count,
        'markdown_cells': md_count
    }

for nb in ['FedRAGFaiss.ipynb', 'RagBenchEvaluation.ipynb', 'SemanticChunking.ipynb']:
    print('Notebook metrics (taken from file):', notebook_overview(nb))
```
