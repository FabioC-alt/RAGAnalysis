```python
from transformers import pipeline
import time

model = "NextGLab/ORANSight_LLama_8B_Instruct"

# Example query

input_query = input("Ask a question about O-RAN: ")


messages = [
    {"role": "system", "content": "You are an O-RAN expert assistant."},
    {"role": "user", "content": input_query},
]

# Load the model
chatbot = pipeline("text-generation", model=model)
start_gen = time.perf_counter()
result = chatbot(messages)

End_gen = time.perf_counter()

print(f"Generation time: {End_gen - start_gen:.2f} seconds")
print(result)

```

```python
import ollama
import time

# Explicitly using the 8B parameter version
model_name = "llama3.1:8b"

# Example query
input_query = input("Ask a question about O-RAN: ")

messages = [
    {"role": "system", "content": "You are an assistant."},
    {"role": "user", "content": input_query},
]

# Start timing
start_gen = time.perf_counter()

# Use the ollama library to handle the request
response = ollama.chat(model=model_name, messages=messages)

# End timing
end_gen = time.perf_counter()

print(f"\nGeneration time: {end_gen - start_gen:.2f} seconds")
print("-" * 30)
print(response['message']['content'])
```
