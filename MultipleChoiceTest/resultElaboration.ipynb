```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df_output = pd.read_csv("newOutput.csv")
df_answer = pd.read_csv("correctAnswer.csv")
df_output.head()
```

```python
## Calculating accuracy of model (FAISS)

matches_faiss = (df_output["FAISS_Letter"]==df_answer["Answer"]).sum()

total = len(df_answer)


percentage = (matches_faiss / total) * 100
```

```python
## Calculating accuracy of model (FAISS)

matches_manual = (df_output["Manual_Letter"]==df_answer["Answer"]).sum()
percentage = (matches_manual / total) * 100

```

```python
## Calculating Latencies (FAISS and Manual)

df_output["Cleaned_FAISS_Latency"] = df_output['Latency FAISS'].str.split().str[0]
df_output["Cleaned_Manual_Latency"] = df_output['Latency Manual'].str.split().str[0]

faiss_mean_latency_value = df_output["Cleaned_FAISS_Latency"].astype(float).mean()
manual_mean_latency_value = df_output["Cleaned_Manual_Latency"].astype(float).mean()


```

```python
matches_cosine = (df_output["Direct_Letter"]==df_answer["Answer"]).sum()
total = len(df_answer)
percentage = (matches_cosine / total) * 100
print(f"Direct Search Accuracy: {percentage:.2f}%")

df_output["Cleaned_Direct_Latency"] = df_output['Latency Direct'].str.split().str[0]
cosine_mean_latency_value = df_output["Cleaned_Direct_Latency"].astype(float).mean()
print(f"Direct Search Mean Latency: {cosine_mean_latency_value:.2f} milliseconds")
```

```python
print(f"FAISS Mean Latency: {faiss_mean_latency_value:.2f} milliseconds")
print(f"FAISS Accuracy: {percentage:.2f}%")
print(f"Manual Mean Latency: {manual_mean_latency_value:.2f} milliseconds")
print(f"Manual Accuracy: {percentage:.2f}%")
```
